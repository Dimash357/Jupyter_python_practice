```python
def multiply(fraction1, fraction2):
    return (fraction1[0] * fraction2[0], fraction1[1] * fraction2[1])

def divide(fraction1, fraction2):
    return (fraction1[0] * fraction2[1], fraction1[1] * fraction2[0])

def smallest(fractions):
    smallest_fraction = fractions[0]

    for fraction in fractions:
        if fraction[0] / fraction[1] < smallest_fraction[0] / smallest_fraction[1]:
            smallest_fraction = fraction

    return smallest_fraction


fraction1 = tuple(map(int, input("Input numbers in tuple, divide them only with space: ").split()))
fraction2 = tuple(map(int, input("Input numbers in tuple, divide them only with space:").split()))
result = multiply(fraction1, fraction2)
print(f"{result[0]}/{result[1]}")

result = divide((5, 3), (10, 3))
print(f"{result[0]}/{result[1]}")

fractions = []
while True:
    value = input("Enter fraction: ")

    if value == "stop":
        break

    fraction = tuple(map(int, value.split()))
    fractions.append(fraction)


result = smallest(fractions)

print(f"{result[0]}/{result[1]}")
```

```python
numbers = []
even = []
odd = []

while True:
    value = input("Input a number: ")
    
    if value == "stop":
        break

    number = int(value)
    numbers.append(number)
    if number % 2 == 0:
        even.append(number)
    else:
        odd.append(number)


print("All numbers:", numbers)
print("Average of all numbers:", sum(numbers) / len(numbers))
print("Sum of all numbers:", sum(numbers))

print("Even numbers:", even)
print("Average of even numbers:", sum(even) / len(even))
print("Sum of even numbers:", sum(even))

print("Odd numbers:", odd)
print("Average of odd numbers:", sum(odd) / len(odd))
print("Sum of odd numbers:", sum(odd))
```

```python
numbers = []

while True:
    value = input("Input a number: ")

    if value == "stop":
        break

    number = float(value)
    if number in numbers:
        continue

    for i in range(len(numbers)):
        if number < numbers[i]:
            numbers.insert(i, number)
            break
    else:
        numbers.append(number)

print(numbers)
```

```python
def analyze_students(names, math_scores, python_scores):
    students = {}

    for number, (name, math, python) in enumerate(
        zip(names, math_scores, python_scores), start=1
    ):
        average = (math + python) / 2

        if average >= 90:
            status = "Excellent"
        elif average >= 75:
            status = "Good"
        else:
            status = "Needs Improvement"

        students[number] = {
            "name": name,
            "math": math,
            "python": python,
            "average": average,
            "status": status
        }

    return students

students = ["Ali", "Bob", "Charlie", "Diana", "Ethan"]
math_scores = [78, 92, 65, 88, 92]
python_scores = [85, 89, 70, 91, 95]

print(analyze_students(students, math_scores, python_scores))
```
