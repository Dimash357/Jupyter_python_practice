Let us plot a line chart where:
    
    i.  x axis will represent weight
    ii. y axis will represent height
    iii. x axis label should be “Weight in kg”
    iv. y axis label should be “Height in cm”    
    v.  colour of the line should be green

```python
import matplotlib.pyplot as plt
import pandas as pd
height=[121.9,124.5,129.5,134.6,139.7,147.3,152.4,157.5,162.6]
weight=[19.7,21.3,23.5,25.9,28.5,32.1,35.7,39.6,43.2]
df=pd.DataFrame({"height":height,"weight":weight})
#Set xlabel for the plot
plt.xlabel('Weight in kg')
#Set ylabel for the plot
plt.ylabel('Height in cm')
#Set chart title:
plt.title('Average weight with respect to average height')
#plot using marker'-*' and line colour as green
plt.plot(df.weight,df.height,marker='*',markersize=10,color='green',linewidth=2, linestyle='--')
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

rng = np.random.default_rng(42)
n = 200

sales_data = pd.DataFrame({
    "date": pd.date_range(
        start="2025-01-01",
        periods=n,
        freq="D"
    ),
    "region": rng.choice(
        ["North", "South", "East", "West"],
        size=n
    ),
    "product": rng.choice(
        ["Laptop", "Tablet", "Phone", "Monitor"],
        size=n
    ),
    "advertising": rng.normal(500, 120, size=n).round(2),
    "units_sold": rng.poisson(20, size=n),
    "customer_rating": rng.normal(4.0, 0.5, size=n).clip(1, 5).round(1)
})

product_price = {
    "Laptop": 1200,
    "Tablet": 600,
    "Phone": 800,
    "Monitor": 400
}

sales_data["price"] = sales_data["product"].map(product_price)

sales_data["sales"] = (
    sales_data["units_sold"] * sales_data["price"]
    + sales_data["advertising"] * 3
    + rng.normal(0, 1500, size=n)
).round(2)

sales_data["expenses"] = (
    sales_data["sales"] * rng.uniform(0.55, 0.85, size=n)
).round(2)

sales_data["profit"] = (
    sales_data["sales"] - sales_data["expenses"]
).round(2)

# Add several unusual observations
sales_data.loc[[15, 75, 150], "sales"] *= 2.5
sales_data.loc[[15, 75, 150], "profit"] *= 2.5

sales_data.head()
```

```python
numeric_columns = ["advertising", "units_sold", "price", "sales", "expenses", "profit", "customer_rating"]
correlation_matrix = sales_data[numeric_columns].corr()

fig, axes = plt.subplots(2, 2, figsize=(15, 11))

# A. Advertising vs. sales: color indicates product; point size indicates units sold
sns.scatterplot(data=sales_data, x="advertising", y="sales", hue="product",
                size="units_sold", sizes=(30, 220), alpha=0.7, ax=axes[0, 0])
axes[0, 0].set_title("A. Advertising vs Sales")
axes[0, 0].set_xlabel("Advertising Spend")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].legend(loc="best", fontsize=8)

# B. Profit distribution
sns.histplot(data=sales_data, x="profit", bins=20, kde=True, color="steelblue", ax=axes[0, 1])
axes[0, 1].axvline(sales_data["profit"].mean(), color="red", linestyle="--", label="Mean")
axes[0, 1].axvline(sales_data["profit"].median(), color="green", linestyle=":", label="Median")
axes[0, 1].set_title("B. Distribution of Profit")
axes[0, 1].legend()

# C. Profit by product
sns.boxplot(data=sales_data, x="product", y="profit", ax=axes[1, 0])
axes[1, 0].set_title("C. Profit by Product")

# D. Numeric variable correlations
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0,
            vmin=-1, vmax=1, ax=axes[1, 1])
axes[1, 1].set_title("D. Correlation Heatmap")

fig.suptitle("Sales Performance Dashboard", fontsize=18, fontweight="bold")
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.savefig("sales_dashboard.png", dpi=160, bbox_inches="tight")
plt.show()

# Quantitative summaries for the interpretation below
print("Advertising-sales correlation:", round(sales_data["advertising"].corr(sales_data["sales"]), 3))
print("Mean profit:", round(sales_data["profit"].mean(), 2))
print("Median profit:", round(sales_data["profit"].median(), 2))
print("Profit by product:")
print(sales_data.groupby("product")["profit"].agg(["median", "std"]).round(2))
print("Sales by product:")
print(sales_data.groupby("product")["sales"].mean().round(2))
print("Top positive correlations:")
import itertools
pairs = sorted(((correlation_matrix.loc[a,b], a,b) for a,b in itertools.combinations(numeric_columns,2)),reverse=True)
print(pairs[:5])
```

### Answers to dashboard questions

**A. Scatter plot**
- Advertising and sales have a weak relationship in this simulated dataset; see the printed Pearson correlation for the exact sign and magnitude.
- Laptops generally generate the highest sales because their unit price is the highest. The product-level means printed above allow verification.
- A few unusually high-sales observations are visible; the dataset deliberately multiplies sales at rows 15, 75 and 150.
- Point size represents **units sold**, while point color represents **product**.

**B. Histogram**
- The distribution is generally right-skewed, with a tail of higher profits.
- The mean and median can be compared using the printed values and the red and green reference lines. The mean may be higher due to large profits.
- Possible extreme profits appear in the right tail, including the deliberately modified observations.
- Increasing the number of bins reveals finer detail but can create a noisier histogram; decreasing bins smooths the distribution and may hide patterns.

**C. Box plot**
- Compare the median lines and the printed product medians to identify the product with the highest median profit (typically Laptop).
- Compare the box heights and the printed standard deviations to identify the product with the greatest variability (typically Laptop).
- Individual points beyond the whiskers indicate possible outliers; check each product's displayed points.
- The **box** covers the interquartile range (25th–75th percentiles), the **center line** is the median, **whiskers** generally extend to the furthest values within 1.5×IQR, and **individual points** beyond them are potential outliers.

**D. Heatmap**
- Sales and expenses show a strong positive correlation, because expenses were generated as a fraction of sales. Sales and profit are also positively related.
- Customer rating has weak relationships with the main financial variables; advertising can also have a relatively weak correlation with sales.
- Customer rating is not strongly related to sales in this simulated data.
- Every diagonal value equals **1** because each variable is perfectly correlated with itself.
- **Correlation does not prove causation**: shared factors, reverse effects, and the data-generation process can produce associations without a causal relationship.
