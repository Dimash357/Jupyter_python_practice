### Task 1. Compute the moving average of window size 3, for the given 1D array.

```python
import numpy as np
np.random.seed(100)
Z = np.random.randint(10, size=10)
# array:  [8 8 3 7 7 0 4 2 5 2]
# moving average:  [ 6.33  6.    5.67  4.67  3.67  2.    3.67  3.  ]
```

```python
result = []

for i in range(len(Z) - 2):
    average = np.mean(Z[i:i + 3])
    result.append(average)

print(np.round(result, 2))
```

### Task 2. Find the index of 5th repetition of number 1 in x.

```python
x = np.array([1, 2, 1, 1, 3, 4, 3, 1, 1, 2, 1, 1, 2])
```

```python
count = 0

for index, number in enumerate(x):
    if number == 1:
        count += 1
        if count == 5:
            print("Index =", index)
            break
```

### Task 4: Weather Data Analysis with NumPy I/O

#### Data generation
Generate synthetic temperature data for 365 days of the year for 5 cities (matrix shape (365, 5)), using random numbers around realistic averages.
Save the data in three formats:

- Binary NumPy format (.npy)

- Text file (.txt)

- CSV file (.csv)

#### Data loading
Load the .npy file back into NumPy.

#### Processing
For each city, compute: 

- Minimum, maximum, mean, and standard deviation of temperature.

- The hottest and coldest day (index of the day).

- Normalize the data (convert to z-scores).

#### Advanced requirement
- Find the day of the year when the most cities had above-average temperature.

- Create a boolean mask for days when all cities were below 0°C.

#### Saving results
Save the computed statistics (per city) into a CSV file with columns: 
City, Min, Max, Mean, Std, HottestDay, ColdestDay. Name it "stats.csv"

Save the normalized data into a .npy file.

```python
import numpy as np
import csv

np.random.seed(100)

cities = ["Almaty", "Astana", "Shymkent", "Aktau", "Karaganda"]
averages = np.array([10, 4, 14, 12, 5])

temperature = np.random.normal(loc=averages, scale=10, size=(365, 5))

np.save("temperatures.npy", temperature)
np.savetxt("temperatures.txt", temperature)
np.savetxt("temperatures.csv", temperature, delimiter=",")

temperature = np.load("temperatures.npy")

minimum = np.min(temperature, axis=0)
maximum = np.max(temperature, axis=0)
mean = np.mean(temperature, axis=0)
std = np.std(temperature, axis=0)

hottest_day = np.argmax(temperature, axis=0)
coldest_day = np.argmin(temperature, axis=0)

normalized = (temperature - mean) / std
np.save("normalized.npy", normalized)

for i, city in enumerate(cities):
    print(city)
    print("Min:", round(minimum[i], 2))
    print("Max:", round(maximum[i], 2))
    print("Mean:", round(mean[i], 2))
    print("Std:", round(std[i], 2))
    print("Hottest day index:", hottest_day[i])
    print("Coldest day index:", coldest_day[i])
    print()

above_average = temperature > mean
city_count = np.sum(above_average, axis=1)
best_day = np.argmax(city_count)

print("Day index with most cities above average:", best_day)
print("Number of cities:", city_count[best_day])

below_zero_mask = np.all(temperature < 0, axis=1)
print("Mask for days when all cities were below 0°C:")
print(below_zero_mask)

with open("stats.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow([
        "City", "Min", "Max", "Mean", "Std", "HottestDay", "ColdestDay"
    ])

    for i, city in enumerate(cities):
        writer.writerow([
            city,
            minimum[i],
            maximum[i],
            mean[i],
            std[i],
            hottest_day[i],
            coldest_day[i]
        ])
```
